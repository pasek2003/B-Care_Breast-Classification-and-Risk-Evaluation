```python
#Library
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

## 1 Understanding data

```python
df= pd.read_csv("../dataset/data.csv")
print("Jumlah data:", df.shape[0])
print("Jumlah fitur:", df.shape[1])
df.head()
df
```

```python
df.info()
```

```python
print("\nJumlah Missing Values per Kolom:")
print(df.isnull().sum())
```

```python
print("\nJumlah duplikat Kolom:")
print(df.duplicated().sum())
```

```python
# hapus kolom yang tidak digunakan

if 'Unnamed: 32' in df.columns:
    df = df.drop('Unnamed: 32', axis=1)

if 'id' in df.columns:
    df = df.drop('id', axis=1)

print("Jumlah data:", df.shape[0])
print("Jumlah fitur:", df.shape[1])

df.head()
```

```python
# counts of unique rows in the 'diagnosis' column
df['diagnosis'].value_counts()
```

```python
#Encoding Label
df['diagnosis'] = df['diagnosis'].map({
'M': 1,
'B': -1
})

df['diagnosis'].value_counts()
```

```python
#Feature dan Label
X = df.drop("diagnosis", axis=1).values
y = df["diagnosis"].values

print("Jumlah data :", len(X))
print("Jumlah fitur :", X.shape[1])
```

## 2. Data Visualization

```python
#Library
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
#Distribusi Kelas (Malignant vs Benign)
plt.figure(figsize=(6,4))

sns.countplot(x=df['diagnosis'])

plt.title("Distribusi Diagnosis Kanker Payudara")
plt.xlabel("Diagnosis (-1 = Benign (Jinak), 1 = Malignant (Ganas))")
plt.ylabel("Jumlah Data")

plt.show()
```

```python
#Histogram Distribusi Fitur
import matplotlib.pyplot as plt

df.drop("diagnosis", axis=1).hist(
    figsize=(22,18),
    bins=20,
    edgecolor="black"
)

plt.suptitle(
    "Distribusi Fitur Dataset Breast Cancer",
    fontsize=20
)

plt.tight_layout(rect=[0,0,1,0.97])

plt.show()
```

```python
plt.figure(figsize=(18,14))

corr = df.corr()

sns.heatmap(
    corr,
    cmap="coolwarm",
    annot=False
)

plt.title("Correlation Matrix Dataset Breast Cancer", fontsize=18)

plt.show()
```

```python
#Fitur Paling Berkorelasi dengan Diagnosis
corr_target = df.corr()['diagnosis'].abs().sort_values(ascending=False)

plt.figure(figsize=(8,10))

sns.barplot(
    x=corr_target.values,
    y=corr_target.index
)

plt.title("Korelasi Fitur terhadap Diagnosis")

plt.xlabel("Nilai Korelasi")

plt.show()
```
